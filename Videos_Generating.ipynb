```python
import cv2
import os
import glob

base_dir = r"D:\Downloads\IP_Project"

frame_dir = os.path.join(
    base_dir,
    "Enhanced_frames"
)

output_video = os.path.join(
    base_dir,
    "Enhanced_frames",
    "Enhanced_video.mp4"
)



frames = sorted(glob.glob(os.path.join(frame_dir, "*.jpg")))

print("Total enhanced frames found:", len(frames))

if len(frames) == 0:
    print("No enhanced frames found.")
    exit()


first_frame = cv2.imread(frames[0])

height, width, layers = first_frame.shape


fps = 30

fourcc = cv2.VideoWriter_fourcc(*'mp4v')

video = cv2.VideoWriter(
    output_video,
    fourcc,
    fps,
    (width, height)
)


for path in frames:

    frame = cv2.imread(path)

    if frame is None:
        print("Skipped:", path)
        continue

    frame = cv2.resize(frame, (width, height))

    video.write(frame)

video.release()

print("===================================")
print("Enhanced video created successfully.")
print("Saved at:")
print(output_video)
print("===================================")
```

```python
import cv2
import numpy as np



video1 = r"D:\Downloads\IP_Project\Road_Damaged_Video.mp4"
video2 = r"D:\Downloads\IP_Project\Enhanced_frames\Enhanced_video.mp4"
video3 = r"D:\Downloads\IP_Project\segmentation_video.mp4"

output_video = r"D:\Downloads\IP_Project\Output_video.mp4"


cap1 = cv2.VideoCapture(video1)
cap2 = cv2.VideoCapture(video2)
cap3 = cv2.VideoCapture(video3)


# VIDEO SIZE

width = 640
height = 360

# Final merged size
final_width = width * 3
final_height = height

fps = 20


fourcc = cv2.VideoWriter_fourcc(*'mp4v')

out = cv2.VideoWriter(
    output_video,
    fourcc,
    fps,
    (final_width, final_height)
)


while True:

    ret1, frame1 = cap1.read()
    ret2, frame2 = cap2.read()
    ret3, frame3 = cap3.read()

    if not (ret1 and ret2 and ret3):
        break

    # Resize videos equally
    frame1 = cv2.resize(frame1, (width, height))
    frame2 = cv2.resize(frame2, (width, height))
    frame3 = cv2.resize(frame3, (width, height))

    # Merge horizontally
    merged = np.hstack((frame1, frame2, frame3))

    # Save output
    out.write(merged)

    # Display
    cv2.imshow("Merged 3 Videos", merged)

    # Press Q to exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap1.release()
cap2.release()
cap3.release()

out.release()

cv2.destroyAllWindows()

print("DONE")
print("Saved at:", output_video)
```

```python
import cv2
import os

# ===============================
# VIDEO PATHS
# ===============================
video1_path = r"D:\Downloads\IP_Project\Output_video Gray.mp4"
video2_path = r"D:\Downloads\IP_Project\Output_video.mp4"

output_path = r"D:\Downloads\IP_Project\merged_2_videos_with_text.mp4"

# Text labels
text1 = "Gray Convesion Output"
text2 = "LAB Color Conversion Output"

# ===============================
# OPEN VIDEOS
# ===============================
cap1 = cv2.VideoCapture(video1_path)
cap2 = cv2.VideoCapture(video2_path)

if not cap1.isOpened():
    print("Cannot open video 1")
    exit()

if not cap2.isOpened():
    print("Cannot open video 2")
    exit()

# ===============================
# VIDEO SETTINGS
# ===============================
fps = int(cap1.get(cv2.CAP_PROP_FPS))

frame_width = 640
frame_height = 480

merged_width = frame_width * 2
merged_height = frame_height

fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_path, fourcc, fps, (merged_width, merged_height))

# ===============================
# MERGE VIDEOS
# ===============================
while True:
    ret1, frame1 = cap1.read()
    ret2, frame2 = cap2.read()

    if not ret1 or not ret2:
        break

    frame1 = cv2.resize(frame1, (frame_width, frame_height))
    frame2 = cv2.resize(frame2, (frame_width, frame_height))

    # Add text background
    cv2.rectangle(frame1, (0, 0), (frame_width, 50), (0, 0, 0), -1)
    cv2.rectangle(frame2, (0, 0), (frame_width, 50), (0, 0, 0), -1)

    # Add text
    cv2.putText(frame1, text1, (160, 35),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)

    cv2.putText(frame2, text2, (160, 35),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)

    # Merge side by side
    merged = cv2.hconcat([frame1, frame2])

    out.write(merged)

    cv2.imshow("Merged 2 Videos", merged)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# ===============================
# RELEASE
# ===============================
cap1.release()
cap2.release()
out.release()
cv2.destroyAllWindows()

print("Merged video saved successfully:")
print(output_path)
```

```python
import cv2
import numpy as np

# =========================================
# VIDEO PATHS
# =========================================
video1_path = r"D:\Downloads\IP_Project\Output_video Gray.mp4"
video2_path = r"D:\Downloads\IP_Project\Output_video.mp4"

output_path = r"D:\Downloads\IP_Project\Gray_LAB_Comparison.mp4"

# =========================================
# TITLES
# =========================================
title1 = "Gray Conversion Output"
title2 = "LAB Color Conversion Output"

# =========================================
# OPEN VIDEOS
# =========================================
cap1 = cv2.VideoCapture(video1_path)
cap2 = cv2.VideoCapture(video2_path)

if not cap1.isOpened():
    print("Cannot open Gray video")
    exit()

if not cap2.isOpened():
    print("Cannot open LAB video")
    exit()

# =========================================
# VIDEO SETTINGS
# =========================================
fps = int(cap1.get(cv2.CAP_PROP_FPS))

frame_width = 1280
frame_height = 360

title_height = 70

output_width = frame_width
output_height = (frame_height + title_height) * 2

fourcc = cv2.VideoWriter_fourcc(*'mp4v')

out = cv2.VideoWriter(
    output_path,
    fourcc,
    fps,
    (output_width, output_height)
)

# =========================================
# PROCESS VIDEOS
# =========================================
while True:

    ret1, frame1 = cap1.read()
    ret2, frame2 = cap2.read()

    if not ret1 or not ret2:
        break

    # Resize videos
    frame1 = cv2.resize(frame1, (frame_width, frame_height))
    frame2 = cv2.resize(frame2, (frame_width, frame_height))

    # =====================================
    # CREATE TITLE BARS
    # =====================================
    title_bar1 = np.ones(
        (title_height, frame_width, 3),
        dtype=np.uint8
    ) * 255

    title_bar2 = np.ones(
        (title_height, frame_width, 3),
        dtype=np.uint8
    ) * 255

    # =====================================
    # ADD TITLES
    # =====================================
    cv2.putText(
        title_bar1,
        title1,
        (30, 45),
        cv2.FONT_HERSHEY_SIMPLEX,
        1.3,
        (0, 0, 0),
        3
    )

    cv2.putText(
        title_bar2,
        title2,
        (30, 45),
        cv2.FONT_HERSHEY_SIMPLEX,
        1.3,
        (0, 0, 0),
        3
    )

    # =====================================
    # COMBINE TITLE + VIDEO
    # =====================================
    top_section = cv2.vconcat([title_bar1, frame1])

    bottom_section = cv2.vconcat([title_bar2, frame2])

    # =====================================
    # FINAL OUTPUT
    # =====================================
    final_output = cv2.vconcat([
        top_section,
        bottom_section
    ])

    # Save video
    out.write(final_output)

    # Display
    cv2.imshow("Gray vs LAB Output", final_output)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# =========================================
# RELEASE
# =========================================
cap1.release()
cap2.release()

out.release()

cv2.destroyAllWindows()

print("Video saved successfully:")
print(output_path)
```
